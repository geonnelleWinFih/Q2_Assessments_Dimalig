## 1. What is Input Validation?

**Input validation** means checking whether data entered by a user meets the requirements of the program before using it.

Examples:
- **Acceptable value:** The answer must be `Yes` or `No`.
- **Data type:** Age must be a whole number.
- **Range:** A score must be from `0` to `100`.
- **Pattern/format:** A phone number must follow a required format.

Validation helps prevent incorrect results and makes programs more reliable.  
The study guide also emphasizes that **output verification** means checking whether the result produced by the program is correct and reasonable.


## 2. Acceptable Value Validation

An acceptable value check makes sure that the user's answer belongs to a list of allowed choices.

### Example
A program only accepts these colors:

`Red`, `Blue`, or `Green`.

We can use the `in` operator to check whether the user's answer is in a list.


````python
allowed_colors = ["Red", "Blue", "Green"]

color = input("Enter your favorite color (Red, Blue, Green): ")

if color in allowed_colors:
    print("You selected:", color)
else:
    print("Invalid color! Please choose from Red, Blue, or Green.")
````

## 3. Data Type Validation

Sometimes the problem is not the value itself, but its **data type**.

For example, if a program asks for an age, it expects a whole number such as `14`, not the word `fourteen`.

Python's `try-except` structure can prevent the program from crashing when text cannot be converted to an integer.


````python
age_input = input("Enter your age: ")

try:
    age = int(age_input)
    print("Your age is:", age)
except ValueError:
    print("Invalid input! Please enter a valid whole number.")
````

## 4. Range Validation

**Range validation** checks whether a numerical value is within an allowed minimum and maximum.

Examples from the lesson:
- Age: `1–120`
- Score: `0–100`
- Percentage: `0–100`

A range check often uses comparison operators such as:

```python
minimum <= value <= maximum
```


````python
score = 85

if 0 <= score <= 100:
    print("Valid score")
else:
    print("Invalid score")
````

## 5. Pattern / Format Validation

Pattern validation checks whether text follows a required format.

For example, the lesson uses a phone number format like:

`09XX-XXX-XXXX`

For this beginner activity, we will use simple string checks instead of regular expressions.


````python
phone = input("Enter your phone number: ")

if phone.startswith("09") and phone.count("-") == 2:
    print("Basic format check passed.")
else:
    print("Invalid format.")
````

## 7. Output Verification

**Output verification** means checking whether the program's result matches what you expect.

For example, if a program calculates:

`20 + 30`

you should verify that the output is:

`50`

Do not simply assume that a program is correct because it runs without an error.


````python
# Example: verify a calculation

number1 = 20
number2 = 30

total = number1 + number2

print("Calculated total:", total)

# Expected output: 50
expected_total = 50

if total == expected_total:
    print("Output verified!")
else:
    print("Check the program. The output is not what was expected.")
````

# Mini Challenge 🧩

## Score Checker

Create a complete Python program that asks for a student's score.

Your program must:

1. Validate that the input is a number.
2. Validate that the score is between `0` and `100`.
3. Assign the appropriate letter grade:
   - 90–100 → A
   - 80–89 → B
   - 70–79 → C
   - 60–69 → D
   - 0–59 → F
4. Display an appropriate message for invalid input.
5. Test your program using at least **three different inputs**.
6. Verify that the displayed grade matches your expected result.

### Before coding

Write your plan:

**Input:** ______________________________

**Validation:** _________________________

**Processing:** _________________________

**Expected output:** ____________________


````python
# MINI CHALLENGE
# Write your complete solution here.

# PLAN

# Input:
#   A raw string from user input representing a student's score (e.g., "85", "105", "abc").

# Validation:
#   1. Try converting the input to a float to verify it's a number.
#   2. Check if 0 <= score <= 100 to ensure it falls within the valid range.

# Processing:
#   Use an if-elif-else chain to evaluate the score against grade thresholds:
#     90 to 100 -> Grade 'A'
#     80 to 89  -> Grade 'B'
#     70 to 79  -> Grade 'C'
#     60 to 69  -> Grade 'D'
#     0 to 59   -> Grade 'F'

# Expected output:
#   - Valid inputs: Display the numerical score alongside its verified letter grade.
#   - Invalid inputs: Display a clear error message indicating why the input failed.



def check_score(raw_input):
    print(f"Testing input: '{raw_input}'")

    # Step 1: Validate numeric conversion
    try:
        score = float(raw_input)
    except ValueError:
        print("Error: Invalid input! Please enter a numerical score.\n")
        return

    # Step 2: Validate score range (0 - 100)
    if score < 0 or score > 100:
        print("Error: Invalid score! Score must be between 0 and 100.\n")
        return

    # Step 3: Assign appropriate letter grade
    if score >= 90:
        grade = "A"
    elif score >= 80:
        grade = "B"
    elif score >= 70:
        grade = "C"
    elif score >= 60:
        grade = "D"
    else:
        grade = "F"

    # Step 4: Display verified output
    print(f"Output verified! Score: {score} | Grade: {grade}\n")


# ==========================================
# STEP 5 & 6: TEST & VERIFY PROGRAM
# ==========================================

# Test 1: Valid high score (Expected: A)
check_score("95")

# Test 2: Valid passing score (Expected: C)
check_score("74.5")

# Test 3: Out of range input (Expected: Error message)
check_score("105")

# Test 4: Non-numeric input (Expected: Error message)
check_score("eighty")
````
