```python
from datetime import date

# Ask for event details and dates
title = input("Enter event title: ").strip()
year = int(input("Enter year: "))
month = int(input("Enter month: "))
day = int(input("Enter day: "))

# Create a date object
event_date = date(year, month, day)

# Display formatted output under here
print("Formatted Output:")
print(f"Event: {title.upper()}")
print(f"Date: {event_date.strftime('%B')} {event_date.day}, {event_date.year}")
print(f"Year: {event_date.year}")
print(f"Month: {event_date.strftime('%B')}")
print(f"Day: {event_date.day}")
print(f"Weekday: {event_date.strftime('%A')}")

#For Output, please type in proper information.
```
